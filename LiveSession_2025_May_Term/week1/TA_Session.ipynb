### Purpose of the Code Cell

This code cell generates a synthetic sales dataset using random values. The goal is to simulate real world transactional data for analysis, visualization, or practice in data manipulation.

Step by Step Functionality

First, a random seed is set to ensure reproducibility. This guarantees that the randomly generated data will be the same each time the code runs.

A date range covering three months is created. Lists of product names and regions are also defined to represent categorical sales information.

A DataFrame named sales_data is then constructed with 500 rows. Each column is populated using random selection or random number generation:

* Date is randomly chosen from the predefined date range.

* Product and Region are randomly selected from their respective lists.

* Units_Sold contains random integer values representing quantity sold.

* Unit_Price contains random floating point values representing product price.

* Customer_ID contains random customer identifiers.

Finally, a new column called Total_Sales is computed by multiplying Units_Sold with Unit_Price. This represents the total revenue generated per transaction.


````python
np.random.seed(42)  # Set a seed for reproducibility so random results remain the same each run

dates = pd.date_range('2023-01-01', '2023-03-31')  # Create a date range from Jan 1, 2023 to Mar 31, 2023
products = ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard']  # List of product names
regions = ['North', 'South', 'East', 'West']  # List of sales regions

sales_data = pd.DataFrame({  # Create a DataFrame with randomly generated sales data
    'Date': np.random.choice(dates, 500),  # Randomly select 500 dates from the date range
    'Product': np.random.choice(products, 500),  # Randomly select 500 products
    'Region': np.random.choice(regions, 500),  # Randomly select 500 regions
    'Units_Sold': np.random.randint(1, 20, 500),  # Generate 500 random integers between 1 and 19 for units sold
    'Unit_Price': np.random.uniform(100, 2000, 500).round(2),  # Generate 500 random prices between 100 and 2000 rounded to 2 decimals
    'Customer_ID': np.random.randint(1000, 1100, 500)  # Generate 500 random customer IDs between 1000 and 1099
})

sales_data['Total_Sales'] = sales_data['Units_Sold'] * sales_data['Unit_Price']  # Create a new column by multiplying units sold with unit price
````

- **Apply on Series**  When you select a single column/Row (which is a Series) or create a standalone Series, .apply() operates element-wise. The function you pass receives individual data values, one at a time. Example : 
```python
         df['score'].apply(lambda x: x * 10)`  
```

- **Apply on DataFrame**  When you call .apply() on an entire DataFrame, it processes data row-wise or column-wise. The function you pass receives a whole Series representing that entire row or column. DOCs [1](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html), [2](https://www.digitalocean.com/community/tutorials/pandas-dataframe-apply-examples)

- Example of using .apply() on a DataFrame:

```python
         # Finds the maximum value of each column (default is axis=0, which means column-wise)
         df.apply(lambda col: col.max(), axis=0) 

         # Adds values from Column_A and Column_B together for every row
         df.apply(lambda row: row['Column_A'] + row['Column_B'], axis=1)
```


````python
def price_cat(u_price):

  if u_price <300:
    return 'Budget'
  elif u_price >= 300 and u_price < 1000:
    return 'Mid-Range'
  else:
    return 'Premium'

````

````python
Price_Category_sr = sales_data['Unit_Price'].apply(func=price_cat)  ## axis = 1 means row wise
Price_Category_sr
````

````python
sales_data['Price_Category'] = Price_Category_sr 
sales_data
````

**Purpose of the Code Cell**

This code cell generates a synthetic employee dataset for analysis, testing, or demonstration purposes.

**Step by Step Functionality**

First, a random seed is set to ensure that the randomly generated data remains consistent across multiple runs.

Two lists are defined to represent possible departments and job positions.

A DataFrame named employee_data is created with 50 employees. The columns include:

Employee_ID generated as a sequential numeric range.

Name generated dynamically using formatted strings.

Department and Position randomly assigned from predefined lists.

Salary generated as random integers within a realistic salary range.

Join_Date randomly selected from a date range between 2018 and 2023 and converted into datetime format.

Performance_Score generated as random integers between 1 and 100.

````python
# Create sample employee data

np.random.seed(42)
# Set random seed to ensure reproducibility of random values

departments = ['Engineering', 'Marketing', 'Finance', 'HR', 'Sales']
# Define a list of department names

positions = ['Junior', 'Mid', 'Senior', 'Manager', 'Director']
# Define a list of job position levels

employee_data = pd.DataFrame({
    'Employee_ID': range(1001, 1051),
    # Generate sequential employee IDs from 1001 to 1050

    'Name': [f'Employee_{i}' for i in range(1, 51)],
    # Create employee names using list comprehension

    'Department': np.random.choice(departments, 50),
    # Randomly assign one of the departments to each employee

    'Position': np.random.choice(positions, 50),
    # Randomly assign one of the job positions to each employee

    'Salary': np.random.randint(40000, 150000, 50),
    # Generate random salary values between 40000 and 149999

    'Join_Date': pd.to_datetime(
        np.random.choice(pd.date_range('2018-01-01', '2023-01-01'), 50)
    ),
    # Randomly select join dates from a date range and convert to datetime format

    'Performance_Score': np.random.randint(1, 101, 50)
    # Generate random performance scores between 1 and 100
})
````
