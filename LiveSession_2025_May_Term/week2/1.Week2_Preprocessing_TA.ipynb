## What is Preprocessing of Data ? 

Data preprocessing involves several transformations that are applied to the raw data to make it more amenable for learning. It is carried out before using it for model training or predicton.

There are many pre-processing Steps : 

*   **Data cleaning**
    *   Data Imputation : meaningful replacement of missing values
    *   Duplicate removal : removing duplicate rows from the dataset
    *   Outlier removal : removing outliers from the dataset
    *   Data type conversion : converting data types of features to appropriate types
    
* **Feature engineering** :  is the process of using domain knowledge to extract features form raw data . These features can be used to improve the performance of machine learning algorithms. It is a crucial step in the machine learning pipeline, as the quality and relevance of features can significantly impact the model's predictive power.
    *   **Feature transformation** : `Scaling`, `categorical encoding`,`discretization of continuous features` , `custom transformers` etc.
        *   *Feature scaling* : meaningful scaling of features to a specific range so that they can be compared on the same scale without giveing high importance to any  feature . 
        *   *Encoding categorical features* 
        *   *Discretization of continuous features*
        *   *Custom Transformers*
        *   *Composite Transformers*
            * Apply transformation to diverse features
            * TargetTransformedRegressor
    *  **Feature Construction** : Creating new features from existing features.
        *   *Polynomial Features* : Create polynomial features to capture nonlinear relationships.
        *   *Domain knowledge based feature construction* : Creating new features based on domain knowledge .
            * for Example : area (newfeature)  =  length * breadth

    *   **Feature Selection** : selecting the most useful existing features.
        * Filter based feature selection
        * Wrapper based feature selection

    *   **Feature Extraction**  : Creating completly new features from existing ones. 
        * PCA

> One important nuance: Some textbooks treat feature engineering as a broader, separate stage that includes preprocessing, while others treat it as a part of preprocessing. `So the hierarchy above is a useful learning framework, not a universal rule`.

> Remember:` Data preprocessing` prepares the data; `feature engineering` focuses on making the features more useful for the ML model.

The transformations in a machine learning workflow are applied in a specific order, which can be defined using Scikit-learn's Pipeline. A pipeline can combine preprocessing steps with a final machine learning estimator.

Different types of features often require different transformations. For example, numerical features may require missing-value imputation and scaling, whereas categorical features may require imputation and encoding. Scikit-learn's ColumnTransformer allows us to apply different transformations to different columns or groups of columns and combine their outputs into a single feature matrix.

`FeatureUnion` serves a different purpose: it applies multiple transformers to the same input in parallel and concatenates their outputs into a single feature matrix.

Scikit-learn also provides tools to visualize pipelines, making it easier to understand the sequence of transformations and how the data flows through the machine learning workflow.

## Handling Missing Values Using Median Imputation

In machine learning, missing values must be handled before training a model because most algorithms cannot work with NaN values.

Median imputation is a technique where missing values are replaced with the median of the column.

Why median is used:

  * It is robust to outliers.

  * It works well for skewed numerical data.

  * It does not get affected heavily by extreme values unlike the mean.

This technique is part of the data preprocessing step, which is essential before model training.

```python
# Replace missing values in column 'A' with the median value of column 'A'
# df['A'].median() calculates the median of all non missing values in column 'A'
# fillna() replaces all NaN values with the computed median
new_A = df['A'].fillna(df['A'].median())
```

## SimpleImputer in Machine Learning

SimpleImputer is a preprocessing utility from sklearn.impute used to handle missing values in datasets.

**Core Concept: Imputation**

Imputation is the process of replacing missing data with substituted values so that machine learning models can work properly.

Key components:

  * missing_values
  
    Specifies what should be considered as missing data. Commonly np.nan.

  * strategy
  
    Determines how missing values are filled. Common strategies include:

    * mean for numerical data

    * median for numerical data

    * most_frequent for categorical or numerical data

    * constant for filling with a fixed value

Why use SimpleImputer instead of manual fillna:

* It integrates smoothly into machine learning pipelines.

* It ensures consistent preprocessing during both training and testing.

* It avoids data leakage when used correctly with train test split.

* This is part of the data preprocessing stage in machine learning workflows

```python
# Create an instance of SimpleImputer
# missing_values=np.nan specifies that NaN values should be treated as missing
# strategy='median' indicates that missing values will be replaced with the median of the column
si_A = SimpleImputer(missing_values=np.nan, strategy='median')
```

```python
si_B = SimpleImputer(missing_values=np.nan, strategy='mean')
```

## fit_transform in Machine Learning

**Core Concept: fit_transform**

The fit_transform method combines two steps into one:

1. fit learns the required parameters from the data

2. transform applies the learned parameters to the same data

In this case:

* fit computes the median of column 'B'

* transform replaces missing values using that median

When to use fit_transform:

* During training data preprocessing

* When you want a shorter and cleaner implementation

When not to use it:

* On test data

Correct workflow:

* Use fit or fit_transform on training data

* Use only transform on test data

This ensures there is no data leakage and the model remains unbiased.

```python
# Fit the SimpleImputer on column 'B' and immediately transform it
# df[['B']] selects column 'B' as a two dimensional DataFrame
# fit_transform() first calculates the median of column 'B'
# Then it replaces all missing values in column 'B' with the learned median
# The output is returned as a NumPy array
si_B.fit_transform(df[['B']])
```

## KNN Imputation

**Core Concept: K Nearest Neighbors Imputation**

KNNImputer is a preprocessing technique from sklearn that fills missing values using the K Nearest Neighbors algorithm.

How it works:

1. For a data point with a missing value, the algorithm finds the K most similar rows based on other available features.

2. It calculates the average value from those nearest neighbors.

3. The missing value is replaced with that average.

In this case:

* n_neighbors=3 means the algorithm will look at the 3 closest rows.

* The missing value will be filled using the mean of those 3 neighbors.

Why use KNN Imputation:

* It preserves relationships between features.

* It can perform better than mean or median imputation when features are correlated.

**Important Note:**

**KNNImputer works based on distance calculations, so feature scaling is often recommended before applying it.**

```python
# Create an instance of KNNImputer
# n_neighbors=3 means that 3 nearest data points will be used to estimate missing values
# The imputer will replace missing values based on the average of the nearest neighbors
knn_A = KNNImputer(n_neighbors=3)
```

## One Hot Encoding vs Ordinal Encoding

**Core Concept: Encoding Categorical Variables**

Machine learning models require numerical input. Categorical features must therefore be converted into numbers.

**OneHotEncoder:**

1. Creates a separate binary column for each category.

2. If a column has 3 categories, it becomes 3 new columns.

3. No artificial order is introduced.

4. Suitable for nominal categories such as colors or cities.

Example:
Red, Blue, Green becomes:

Red → 1 0 0

Blue → 0 1 0

Green → 0 0 1

**OrdinalEncoder:**

1. Assigns an integer to each category.

2. Categories are mapped to numbers like 0, 1, 2.

3. Suitable only when categories have a natural order.

Example:
Low, Medium, High becomes:

Low → 0

Medium → 1

High → 2

**Important Note:**

**Using OrdinalEncoder on purely nominal data may introduce unintended ordering relationships, which can negatively affect certain models.**

```python
# Create an instance of OneHotEncoder
# sparse_output=False ensures the output is returned as a dense NumPy array instead of a sparse matrix
# OneHotEncoder converts categorical values into binary columns
ohe = OneHotEncoder(sparse_output=False)

# Create an instance of OrdinalEncoder
# OrdinalEncoder converts categorical values into integer encoded labels
# Each unique category is assigned a numerical value
oe = OrdinalEncoder()
```

## Sparse vs Dense Matrix in One Hot Encoding

**Core Concept: Sparse and Dense Matrix Representation**

When using OneHotEncoder, the transformed output can be stored in two formats: sparse or dense.

**Dense Matrix:**

* Stores every value including zeros.

* Represented as a regular NumPy array.

* Takes more memory when there are many zeros.

* Easier to view and manipulate directly.

Example dense output:

1 0 0

0 1 0

0 0 1

**Sparse Matrix:**

* Stores only non zero values.

* Saves memory when the matrix contains many zeros.

* Common in one hot encoding because most values are zero.

* More efficient for high dimensional categorical features.

**Why sparse is useful in One Hot Encoding:**

If a feature has many categories, one hot encoding creates many columns.
For each row, only one column will contain 1 and the rest will be 0.
This results in a matrix filled mostly with zeros.

Using sparse representation:

* Reduces memory usage.

* Improves computational efficiency for large datasets.

In sklearn:

* sparse_output=True returns a sparse matrix.

* sparse_output=False returns a dense NumPy array.

When to use dense:

* Small datasets.

* When easier inspection is needed.

* When the next processing step does not support sparse input.

When to use sparse:

* Large datasets.

* High cardinality categorical features.

* Memory efficiency is important.

## Ordered Encoding with OrdinalEncoder

**Core Concept: Ordinal Encoding with Explicit Category Order**

OrdinalEncoder converts categorical values into integers.
By default, it assigns numbers based on alphabetical order.

However, in many real world problems, categories have a natural ranking.

**Example:**

low < medium < high

If we do not specify the order, the encoder might assign values incorrectly based on alphabetical sorting.

By defining:

categories=`[['low','medium','high']]`

we explicitly control the mapping:

low -> 0

medium -> 1

high -> 2

Why this is important:

Preserves the real world order of categories.

Allows models such as linear regression or logistic regression to interpret increasing magnitude correctly.

Prevents incorrect ordinal relationships.

Use this approach only when categories truly have a meaningful order.

```python
# Create an OrdinalEncoder with a predefined category order
# categories=[['low','medium','high']] explicitly defines the ranking order
# 'low' will be encoded as 0
# 'medium' will be encoded as 1
# 'high' will be encoded as 2
# This ensures the encoder respects the logical order of the categories
oe1 = OrdinalEncoder(categories=[['low','medium','high']])
```

## Feature Scaling in Machine Learning

**Core Concept: Feature Scaling**

Feature scaling ensures that numerical features are on a similar scale so that machine learning models perform efficiently and fairly.

1. StandardScaler

    Formula:

    **z = (x − mean) / standard deviation**

* Transforms data to have mean 0 and standard deviation 1.

* Useful for algorithms based on distance or gradient descent.

* Commonly used in logistic regression, SVM, and KNN.

2. MinMaxScaler

    Formula:

    **x_scaled = (x − min) / (max − min)**

* Scales data into a fixed range, typically 0 to 1.

* Preserves the shape of the original distribution.

* Sensitive to outliers.

3. MaxAbsScaler

    Formula:

    **x_scaled = x / max absolute value**

* Scales values to range between −1 and 1.

* Does not shift or center data.

* Works well with sparse matrices because it does not destroy sparsity.

**Why scaling matters:**

**Many algorithms such as KNN, SVM, and gradient based models are sensitive to feature magnitude. Without scaling, features with larger values can dominate the model.**

```python
# Create an instance of StandardScaler
# StandardScaler standardizes features by removing the mean and scaling to unit variance
# Each feature will have mean equal to 0 and standard deviation equal to 1
ss = StandardScaler()

# Create an instance of MinMaxScaler
# MinMaxScaler scales features to a fixed range, usually between 0 and 1
# It rescales the data based on the minimum and maximum values of each feature
mms = MinMaxScaler()

# Create an instance of MaxAbsScaler
# MaxAbsScaler scales each feature by dividing by its maximum absolute value
# It preserves zero entries and is useful for sparse data
mas = MaxAbsScaler()
```
