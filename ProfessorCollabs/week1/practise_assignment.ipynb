MetaData About the Flight Data Set:
- `Airline`:The name of the airline 
- `Source`: The source from which the service begins 
- `Destination`: The destination where the service ends 
- `Route`: Route the flight took. 
- `Dep_Time`: The time when the journey starts from the source. 
- `Arrival_Time`: Time of arrival at the destination. 
- `Duration`: Total duration of the flight. 
- `Total_Stops`: Total stops between the source and destination. 
- `Additional_Info`: Additional information about the flight 
- `Price`: The price of the ticket ▶▶▶ Target 
- `Month`: Month of journey. 
- `WeekDay`: Day at which journey started. 
- `Day`: Date of the start of journey.

```python
df["duration_seconds"] = pd.to_timedelta(
    df['Duration'].str.replace('h', ' hours').str.replace('m', ' minutes') , 
    errors='coerce'
).dt.total_seconds()
```

```python
## How many flights started in the Morning and arrived the destination at Evening?

"""
Apply the following functions to the columns Dep_Time and Arrival_Time:

Transform the values in the 'dep_time' and 'arrival_time' columns to represent the hour component. For instance, if an entry is 10:05 June 13 or 10:05, the corresponding value should be 10. 
Then convert the time into four categories as follows:
5 <= hour < 12 = Morning 
12 <= hour < 17 = Afternoon 
17 <= hour < 20 = Evening 
20 <= hour < 5 = Night


"""
# Departure time
df['Dep_Time'] = pd.to_datetime(
    df['Dep_Time'],
    format='%H:%M',
    errors='coerce'
)

# Arrival time: if the column has mixed formats,
# use format='mixed' with dayfirst=True when appropriate.
df['Arrival_Time'] = pd.to_datetime(
    df['Arrival_Time'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)

df[(df["Dep_Time"].dt.hour >= 5) & (df["Dep_Time"].dt.hour < 12) & (df["Arrival_Time"].dt.hour >= 17) & (df["Arrival_Time"].dt.hour < 20)].shape
```

```python
'''
Encode the values of column 'WeekDay' as follows:
Weekends (Sunday, Saturday) = 1
all remaining five days = 0
What is the most frequent (mode) WeekDay?

'''

def encode_weekday(row):
    if row['WeekDay'] in ['Saturday', 'Sunday']:
        return 1
    else:
        return 0
    

df["encoded_WeekDay"] = df.apply(encode_weekday, axis=1)
df.head()
```
