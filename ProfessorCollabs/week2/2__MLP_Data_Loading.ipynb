# Introduction

The objective of this colab is to demonstrate `sklearn` dataset API.

Recall that it has three APIs:
1. Loaders (`load_*`) load small standard datasets bundled with `sklearn`.
2. Fetchers (`fetch_*`) fetch large datasets from the internet and loads them in memory.
3. Generators (`generate_*`) generate controlled synthetic datasets.

Loaders and fetchers return a `bunch` object which has a `data` attribute for the feature matrix and a `target` attribute for the label vector (or matrix). Generators return a tuple of feature matrix and label vector (or matrix). but if we pass `return_X_y=True` to the Loaders or Fetchers, it will also return a tuple of feature matrix and label vector (or matrix).

## Loading iris dataset

````python
from sklearn.datasets import load_iris
data = load_iris()
data # Bunch object with data, target, feature_names, target_names, DESCR, filename
````

````python
X, y = load_iris(return_X_y=True, as_frame=True)
````

This returns a `Bunch` object `data` which is a dictionary like object with the following attributes:
* `data`, which has the feature matrix.
* `target`, which is the label vector
* `feature_names` contain the names of the features.
* `target_names` contain the names of the classes.
* `target` gives the class labels as array of integers for each sample. The mapping of integer to class name is given by **target_names**.
* `DESCR` has the full description of dataset.
* `filename` has the path to the location of data.

Let's generate 100 samples with 5 features for multiple regression problem with 5 outputs. **Multi-output regression**

- In Multi-output regression, the model predicts multiple target variables simultaneously.
- Unlike traditional single-output regression—which forecasts a single target variable (e.g., predicting a house's price)—multiple-output regression forecasts a vector of values. 
-For a given input features vector `X = (x_1, x_2, ..., x_p)`, the model outputs a set of continuous targets `Y = (y_1, y_2, ..., y_k)` at the same time

## `make_multilabel_classification`

This function helps us generating a random multi-label classification problem.

````python
from sklearn.datasets import make_multilabel_classification
# ?make_multilabel_classification
````

Let's generate a multilabel classification problem with 100 samples, 10 features, 3 labels and on an average 2 labels per example.

````python
X, y = make_multilabel_classification(n_samples=100, n_features=10, n_classes=3, n_labels=2 , random_state=42)

## n_samples: The number of samples (data points) to generate.
## n_features: The number of features (input variables) for each sample.
## n_classes: The number of possible classes (labels) for each sample.
## n_labels: The average number of labels assigned to each sample. ie ;  if n_labels=2, then on an average each sample will have 2 labels assigned to it.
##  Example: if  n_classes=3, n_labels=2
            # Then for each data point the label will be a vector [0, 1, 1] or [1, 0, 1] or [1, 1, 0] etc. (3 classes and each data point will have 2 labels assigned to it)
            # Each index of the vector corresponds to a class, and a value of 1 indicates that the sample belongs to that class, while a value of 0 indicates that it does not.
````

First of all, let's examine shapes of feature matrix and label vector.

````python
X.shape
````

````python
y.shape
````

````python
y[:5] 
````

[1 ,1 ,0 ] represents that the first sample has label 0 and label 1, but not label 2. The second sample has label 0 and label 2, but not label 1. The third sample has only label 1, and so on.

## `make_blobs`

`make_blobs` enables us to generate random data for clustering.

````python
from sklearn.datasets import make_blobs
# ?make_blobs
````

Let's generate a random dataset of 10 samples with 2 features each for clustering.

````python
X, y = make_blobs(n_samples=100, centers=3, n_features=2, random_state=42)
print ("Feature matrix shape:", X.shape)
print ("Label shape:", y.shape)
labels = [(X[i, 0], X[i, 1]) for i in range(X.shape[0])]
````

We can find the cluster membership of each point in `y`.

````python
y
````

````python
import numpy as np

plt.scatter(X[:, 0], X[:, 1], c=y, cmap="viridis", marker="h", alpha=0.7)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Synthetic Clustering Dataset")
````

Lets visualize the 3d clusters 

````python
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-darkgrid')
# plt.style.available
````

````python

X , y = make_blobs( n_samples=100, centers=3, n_features=3, random_state=42 )
````

````python
fig , axes = plt.subplots(1 ,1)
axes = fig.add_subplot(111, projection='3d')
axes.scatter(X[:, 0], X[:, 1], X[:, 2], c=y, cmap="viridis", marker="o", alpha=0.7)
axes.set_xlabel("Feature 1")
axes.set_ylabel("Feature 2")
axes.set_zlabel("Feature 3")
axes.set_title("3D Synthetic Clustering Dataset")
````
