#### Encoding Categorical Features

Types of Encoding:
1. One Hot Encoding 
        - it creates a new binary column for each category of the categorical feature. Each row will have a 1 in the column corresponding to its category and 0s in all other columns. This is mainly used for nominal categorical features. (no ordering between the categories)

2. ordinal encoding
        - it assigns a unique integer to each category of the categorical feature. Each row will have the integer corresponding to its category. This is mainly used for ordinal categorical features. (ordering between the categories)

3. Label Encoding
        - it assigns a unique integer to each category of the categorical feature. Each row will have the integer corresponding to its category. This is mainly used for ordinal categorical features. (ordering between the categories)

4. Lable Binarization
        - it is used to convert a multi-class target variable into a binary format. It creates a binary column for each class of the target variable. Each row will have a 1 in the column corresponding to its class and 0s in all other columns.

5. MultiLabel Binarization
        - it is used to convert a multi-label target variable into a binary format. It creates a binary column for each label of the target variable. Each row will have a 1 in the column corresponding to its label and 0s in all other columns.

```python
data = {
    "Name" : ["Alice", "Bob", "Charlie", "David", "Eve" , "Frank"],
    "city": ["Kolkata", "Mumbai", "Delhi", "Kolkata", "Mumbai", "Delhi"],  # Nominal -> OneHotEncoder
    "education": ["PhD", "Bachelor", "Master","High School", "PhD", "Bachelor"],  # Ordinal -> OrdinalEncoder
    "purchased": ["No", "Yes", "Yes", "No", "Yes" , "No"],  # Target -> LabelEncoder
}

df = pd.DataFrame(data)
df
```

```python
from sklearn.preprocessing import OrdinalEncoder , OneHotEncoder , LabelEncoder

ohe = OneHotEncoder(sparse_output=False).set_output(transform="pandas")
ohe.fit_transform(df[["city"]])
```

```python
oe = OrdinalEncoder(categories=[["High School", "Bachelor", "Master", "PhD"]]).set_output(transform="pandas")
oe.fit_transform(df[["education"]])
```

`LabelEncoder` vs. `OrdinalEncoder`

| Feature | `LabelEncoder` | `OrdinalEncoder` |
| :--- | :--- | :--- |
| **Intended Purpose** | Target variable (`y`) only | Input features (`X`) |
| **Expected Input Shape** | **1D** (`df["purchased"]`) | **2D** (`df[["education"]]`) |
| **Number of Columns** | Single column only | One or multiple columns at once |
| **Custom Ranking Order** | **No** — always sorts alphabetically (`0, 1, 2...`) | **Yes** — via `categories=[["Low", "Med", "High"]]` |
| **Works in `Pipeline` / `ColumnTransformer`?** | **No** (throws a shape/argument error) | **Yes** (built specifically for pipelines) |
| **Handling Unseen Categories** | Raises a `ValueError` | Can assign a fallback value (`handle_unknown="use_encoded_value"`) |
| **Output Data Type** | Integers (`0, 1, 2`) | Floats (`0.0, 1.0, 2.0`) |

#### Column Transformer


The ColumnTransformer is a feature transformer from scikit-learn that allows different
columns or column subsets of the input dataset to be transformed separately and the features
generated by each transformer to be concatenated into a single feature space. This is
particularly useful for datasets that contain various types of data requiring different
preprocessing steps, such as numerical data that needs scaling and categorical data that needs
to be encoded.


What is allowed in Column Transformer?
Transformers, Function Transformers, Custom Transformers, Pipelines --> Touched Later in the noteBook, Feature Unions

```python

data = {
    "age": [25, 32, 45, 28, 38, 52, 23, 41, 35, 29],

    "annual_income": [
        450000, 850000, 1200000, 600000, 950000,
        1500000, 350000, 1100000, 750000, 550000
    ],

    "credit_score": [
        720, 680, 790, 650, 740,
        810, 690, 760, 700, 730
    ],

    "city": [
        "Mumbai", "Delhi", "Kolkata", "Mumbai",
        "Chennai", "Delhi", "Kolkata", "Mumbai",
        "Chennai", "Delhi"
    ],

    "gender": [
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],

    "education": [
        "Bachelor", "Master", "High School", "PhD",
        "Bachelor", "PhD", "Master", "High School",
        "Bachelor", "Master"
    ],

    "satisfaction": [
        "Low", "High", "Medium", "High", "Low",
        "Medium", "High", "Low", "Medium", "High"
    ],

    "customer_id": [
        1001, 1002, 1003, 1004, 1005,
        1006, 1007, 1008, 1009, 1010
    ],

    "account_age_months": [
        12, 36, 60, 18, 48, 84, 6, 72, 30, 24
    ],

    "area": [
        "Urban", "Urban", "Rural", "Suburban", "Urban",
        "Rural", "Suburban", "Urban", "Rural", "Suburban"
    ]
}

df = pd.DataFrame(data)
df
```


Practice Questions (Solve Using a Single ColumnTransformer)

1. Standard Scaling (age): How do you standardize the age column so it has a mean of 0 and standard deviation of 1 using StandardScaler?
2. Min-Max Scaling (annual_income): How do you rescale annual_income to fall strictly between 0 and 1 using MinMaxScaler?
3. Discretization / Binning (credit_score): How do you group credit_score into 3 equal-width ordinal bins (0, 1, 2) using KBinsDiscretizer?
4. One-Hot Encoding (city): How do you convert the nominal city column into binary dummy columns without returning a sparse matrix (sparse_output=False)?
5. Binary One-Hot Encoding (gender): How do you encode gender into a single binary column by dropping the first category (drop='first')?
6. Custom Ordinal Encoding (education): How do you map education into ordered integers following the exact hierarchy ['High School', 'Bachelor', 'Master', 'PhD']?
7. Custom Ordinal Encoding (satisfaction): How do you map satisfaction into ordered integers following the exact hierarchy ['Low', 'Medium', 'High']?
8. Dropping Columns (customer_id): How do you explicitly remove customer_id inside the transformers list using the 'drop' keyword?
9. Passing Remaining Columns (account_age_months): How do you ensure account_age_months stays untouched in the final output without listing it inside transformers?
10. Inspecting Output: How do you fit, transform df, and convert the resulting NumPy array back into a labeled pandas DataFrame using get_feature_names_out()?


```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, KBinsDiscretizer
ct = ColumnTransformer(
    [
    ("Std_scaling" , StandardScaler() , ["age"]),
    ("MinMax_scaling" , MinMaxScaler() , ["annual_income"]),
    ("Discritization" , KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='uniform') , ["credit_score"]),
    ("OHE_city" , OneHotEncoder(sparse_output=False) , ["city"]),
    ("OHE_gender" , OneHotEncoder(sparse_output=False) , ["gender"]),
    ("Ordinal_encod_Education" , OrdinalEncoder(categories=[["High School", "Bachelor", "Master", "PhD"]]) , ["education"]),
    ("Ordinal_encod_Satisfaction" , OrdinalEncoder(categories=[["Low", "Medium", "High"]]) , ["satisfaction"]),
    ("drop_customer_id" , "drop" , ["customer_id"]),

] , remainder="passthrough").set_output(transform="pandas")
```

```python
# 1. Min-Max Scaler for ["CRIM" ,"ZN" , "POLINDEX" ,"DIS" ,"HIGHWAYCOUNT" ,"TAX" ,"PTRATIO" ,"IMM" ,"BPL" ] {column transformer}
# 2. standard scaling for ["INDUS" ] (column transformer)
## ___ column transformer for 3 pipes__
# 3. Impute with median for ["RM" ] -->  Min-Max Scaler for ["RM"] {pipeline}
# 4. Impute with mean for ["AGE" ] -->  Min-Max Scaler for ["AGE"] {pipeline}
# 5. Impute with most frequent value for ["RIVERSIDE" ] --> OHE for ["RIVERSIDE"] {pipeline}
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

rm_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", MinMaxScaler()),
])

age_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", MinMaxScaler()),
])

river_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("ohe", OneHotEncoder(sparse_output=False)),
])

preprocessor = ColumnTransformer(
    transformers=[
        ("step_minmax", MinMaxScaler(), ["CRIM", "ZN"]),
        ("step10_indus_ss", StandardScaler(), ["INDUS"]),
        (
            "step5_11_minmax",
            MinMaxScaler(),
            [
                "POLINDEX",
                "DIS",
                "HIGHWAYCOUNT",
                "TAX",
                "PTRATIO",
                "IMM",
                "BPL",
            ],
        ),
        ("step12_rm", rm_pipe, ["RM"]),
        ("step13_age", age_pipe, ["AGE"]),
        ("step14_riv", river_pipe, ["RIVERSIDE"]),
    ],
    remainder="drop",  # Automatically drops CATEGORY (and anything else not listed)
    verbose_feature_names_out=False
).set_output(transform="pandas")

preprocessor
```

Practising Cross validation for the the Errors

- syntax for cross_val_score : `cross_val_score(estimator, X, y, cv=..., scoring="..")`
- estimator : estimator object implementing 'fit'
    The object to use to fit the data.
- X : {array-like, sparse matrix} of shape (n_samples, n_features)
    The data to fit. Can be for example a list, or an array.
- y : array-like of shape (n_samples,) or (n_samples, n_outputs),             default=None
    The target variable to try to predict in the case of
    supervised learning.
- scoring : str or callable, default=None
    A str (see model evaluation documentation) or
    a scorer callable object / function with signature
    ``scorer(estimator, X, y)`` which should return only
    a single value.
- cv : int, cross-validation generator or an iterable, default=None
    Determines the cross-validation splitting strategy.
    Possible inputs for cv are:

    - `None`, to use the default 5-fold cross validation,
    - int, to specify the number of folds in a `(Stratified)KFold`,
    - :term:`CV splitter`,
    - An iterable that generates (train, test) splits as arrays of indices.

    - For `int`/`None` inputs, if the estimator is a classifier and `y` is
    either binary or multiclass, :class:`StratifiedKFold` is used. In all
    other cases, :class:`KFold` is used. These splitters are instantiated
    with `shuffle=False` so the splits will be the same across calls.

```python
from sklearn.model_selection import cross_val_score , KFold , ShuffleSplit

KFold = KFold(n_splits=5, shuffle=True, random_state=42)
score = cross_val_score(ridge_model, x_train, y_train, cv=KFold, scoring='r2')
display(score)
display(score.mean())
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import SGDRegressor
```
